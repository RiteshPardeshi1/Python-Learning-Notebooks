```python
#find the length in a string
```

```python
def string_length(s):
    count = 0
    for x in s:
        count += 1
    return count
```

```python
#Ritesh write a code and which gives us odd and even numbers
```

```python
def odd_even(list):
    even = [x for x in list if x % 2 == 0]
    odd = [x for x in list if x % 2 != 0]
    return even, odd
```

```python
odd_even(l1)
```

```python
# Check if the number is positve negative zero
```

```python
def check_number(num):
    if num  > 0:
        return "positive"
    elif num < 0:
        return "negative"
    else:
        return "zero"
```

```python
check_number(0.1)
```
