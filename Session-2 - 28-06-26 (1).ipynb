## f-strings for String formatting  


- Introduced in Python 3.6.
- f-strings are the current standard. 
- They are the most readable, fastest, and easiest to use. 
- You simply prefix a string with f and place variables directly inside curly braces {}.

`1. Syntax Basics`

Place an f or F before your quotation marks. Inside the string, use curly braces {} to wrap the variables or expressions you want to include.

Example: f"My name is {name}"

`2. Dynamic Insertion` 

Python automatically replaces the content inside the {} with the actual value of the variable at that moment.

Example: If age = 25, then f"I am {age}" becomes "I am 25".


`3. Run Code Directly` 

You aren't limited to just variables; you can put math, function calls, or method calls directly inside the curly braces.

Example: f"2 + 2 = {2 + 2}" becomes "2 + 2 = 4".


`4.Format Data on the Fly` 

You can add a colon : inside the braces to format numbers, dates, or strings (like rounding decimals, adding commas, or aligning text).

Example: f"Price: ${49.995:.2f}" becomes "Price: $50.00".

`5. Not Just for Printing` 

While they are popular for print() statements, f-strings create a standard string object. You can store them in variables, write them to files, or send them to a database.

Example: greeting = f"Hello, {user}!"



```python
name = "Alex"

version = 2.1

#print("My name is",name, "and my version is",version)

print(f"Hello, I am {name}, version {version + 1}.")
```

#### 3.1 The Zip Function



- Takes in Two or more tuples (or lists).

- Returns an iterator of tuples. Each tuple contains one element from each input at the corresponding index.

- zip() stops as soon as the shortest input tuple is exhausted.

```python
# Assume you have  two tuples: one for product names and one for their prices. You want to pair them together.


products = ("Laptop", "Mouse", "Keyboard")
prices = (999.99, 25.50, 45.00)

# zip() pairs them up
zipped_data = zip(products, prices)

# Since zip() returns an iterator
result = tuple(zipped_data)

print(result)
```
