## 6. Combining Dictionaries

```python
# Current inventory in our warehouse
warehouse_inventory = {
    "prod_001": {"name": "Stethoscope", "price": 120.00, "stock": 5},
    "prod_002": {"name": "Blood Pressure Monitor", "price": 85.50, "stock": 0}
}

# New shipment arriving today
new_shipment = {
    "prod_002": {"name": "Blood Pressure Monitor", "price": 85.50, "stock": 20},
    "prod_003": {"name": "Pulse Oximeter", "price": 45.00, "stock": 10}
}
```

```python
# 6.1 
# Unpack & Merge: Creates a new dictionary, leaving the original inventory untouched

combined_catalog = {**warehouse_inventory, **new_shipment}

combined_catalog
```

```python
# 6.2
# Union Operator: A clean way to create a new dictionary

# Note: Rightmost dict wins on duplicate keys 

latest_catalog = warehouse_inventory | new_shipment

latest_catalog
```

```python
# 6.3 Update

# An in-place merge that modifies the original warehouse_inventory directly
# This is the fastest way to perform a merge because no new object is created

warehouse_inventory.update(new_shipment)

warehouse_inventory
```

```python
## 8.2 Shallow Copy
```

```python
import copy

# A product entry in an inventory system

product = {
    "name": "Laptop",
    "tags": ["electronics","sale"]  
}
```

```python
# The Trap: Shallow Copy

copy_product = product.copy()
safe_copy = copy.deepcopy(product)

```

```python
safe_copy
```

```python
# If you change the nested list, the original 'product' changes too.

safe_copy["tags"].append("clearance")

```

```python
product
```

```python
safe_copy
```


` Why the Shallow Copy is not the best way to copy`

- .copy() i.e shallow copy, copies the "top-level" container. 
- The specs and tags inside are still pointing to the original memory locations.


```python
## 8.2 Deep Copy
```

```python
# The Fix: Deep Copy

safe_copy = copy.deepcopy(product)

# Now, we can modify the nested structures safely

safe_copy["specs"]["ram"] = "32GB"
safe_copy["tags"].append("new-arrival")

print(product["specs"]["ram"])

print(product["tags"])
```


`deepcopy()` recursively walks through the entire dictionary and creates a truly independent copy of every nested piece.


`Summary of Differences` 

1 - Assignment (b = a): Creates an alias; both names point to the same object.

2 - Shallow Copy (a.copy()): Safely copies the top level, but nested objects (lists/dicts) remain linked.

3 - Deep Copy (deepcopy(a)): Creates a completely independent clone, perfect for nested structures.

# 4 . Mathematical Operations on elements of a SET

```python
# Students enrolled in two different language classes

spanish_class = {"Alice", "Bob", "Charlie", "David"}

french_class = {"Charlie", "David", "Eve", "Frank"}


print(f" Spanish students are {spanish_class}, and French students are {french_class}")
```

```python
#4.1 Union

# 1. Union (|): Combines all unique students from both classes.


all_students = spanish_class | french_class

all_students
```

```python
#4.2 Intersection

# 2. Intersection (&): Finds students enrolled in BOTH classes.

dual_enrolled = spanish_class & french_class

dual_enrolled
```

```python
#4.3 Difference
# 3. Difference (-): Finds students enrolled in Spanish but NOT in French.


spanish_only = spanish_class - french_class

spanish_only
```

```python
#4.4 Symmetric Difference 

# 4. Symmetric Difference (^): Finds students enrolled in one class OR the other, but NOT both (those who are not dual-enrolled).

unique_to_one = spanish_class ^ french_class

unique_to_one

dual_enrolled = spanish_class & french_class

dual_enrolled

print(f" learning only one {unique_to_one}. Learning both {dual_enrolled}")
```

# 5 . SET Operations on elements of a SET

```python
all_required_tools = {"Python", "R", "SQL", "Tableau", "Git"} #a

data_science_team = {"Python", "SQL"}  #b

bioinformatics_team = {"R", "SAS"}   #c 
```

```python
# 1. issubset(): Checks if all elements of one set are contained within another.
# Returns True if data_science_team is a subset of all_required_tools.

is_subset = data_science_team.issubset(all_required_tools)

is_subset
```

```python
# 2. issuperset(): Checks if a set contains all elements of another set.
# Returns True if all_required_tools is a superset of data_science_team.

is_superset = all_required_tools.issuperset(data_science_team)

is_superset
```

```python
# 3. isdisjoint(): Checks if two sets have NO elements in common.
# Returns True if the sets have no shared items (bioinformatics_team vs data_science_team).

is_disjoint = bioinformatics_team.isdisjoint(data_science_team)

is_disjoint
```

### 6.2  Copy

Using .copy() creates a completely independent set. If you modify your new copy, the original list remains untouched.

```python
grocery_list = {"Milk", "Bread", "Eggs"}
my_personal_copy = grocery_list.copy()

my_personal_copy.add("Coffee")
print(grocery_list)    # {'Milk', 'Bread', 'Eggs'} — Unchanged
print(my_personal_copy) # {'Milk', 'Bread', 'Eggs', 'Coffee'}
```

### 6.3 set() Constructor 

The set() constructor creates an independent set from an existing iterable. This is very useful if you want to ensure your modifications don't affect the original source.

```python
grocery_list = {"Milk", "Bread", "Eggs"}

my_personal_copy = set(grocery_list) # New independent set

my_personal_copy.add("Milk")

print(f" The Original {grocery_list}. The duplicate {my_personal_copy}")
```
