### 13. Boolean Expressions

```python
# Find the students based in Mumbai

df[df['City'] == 'Mumbai']
```

```python
# Which students have a Score greater than 90?
df[df['Score'] >90]
```

```python
# Which students have an attendance rate lower than 70?

df[df['Attendance'] < 70]
```

```python
# Which students are not from Bangalore?

df[df['City'] != 'Bangalore']
```

```python
# Which students have an Attendance above 90 AND a Score above 85?

df[(df['Attendance'] > 90) & (df['Score'] > 85)]
```

```python
# Which students have a Score of 95 OR Subjects_Passed equal to 6?

df[(df['Score'] == 95) | (df['Subjects_Passed'] == 6)]
```
