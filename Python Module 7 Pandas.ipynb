```python
# Assume you keep a tab of the number of steps that you walk in a day

daily_steps = [8200, 6100, 11400, 9800, 4200, 15300, 7600]
index = ["Mon", "Tue", "Wed", "Thu", "Fri", "Sat", "Sun"]

# Iterate over all elements using a loop
for day,steps in zip(index, daily_steps):
    print(day,steps)
```
