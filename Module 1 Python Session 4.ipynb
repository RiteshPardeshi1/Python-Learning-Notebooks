```python
# ask user to enter a word and print the count of each word

d = {}

word = input("enter a word")

for i in word:
    if i not in d:
        d[i] = 1
    else:
        d[i] +=1


for k,v in d.items():
    print(k,v)
```

```python
word = "hippopotamus"

found = {}

for i in word:
    if i in "aeiou":
        found.setdefault(i,0)
        found[i] += 1
    else:
        pass

for k,v in found.items():
    print(f"the letter {k} appers {v} times")
```

```python
# Assume we have a dictionary that stores values of inventory as below

inventory = {
    "prod_001": {"name": "Stethoscope", "price": 120.00, "stock": 5},
    "prod_002": {"name": "Blood Pressure Monitor", "price": 85.50, "stock": 0},
    "prod_003": {"name": "Pulse Oximeter", "price": 45.00, "stock": 12},
    "prod_004": {"name": "Digital Thermometer", "price": 20.00, "stock": 8},
    "prod_005": {"name": "First Aid Kit", "price": 35.00, "stock": 0}
}
```

```python
print("prod_001" in inventory)
```

```python
print("prod_001"  not in inventory)
```

```python
print(len(inventory))
```

```python
count = 0 
for item in inventory.values():
    count += item["stock"]
print(f"Total stock across all products: {count}")
```

```python
sum(item["stock"] for item in inventory.values())
```

```python
inventory['prod_001']['stock']
```
