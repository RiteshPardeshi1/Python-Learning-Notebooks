```python
goals = pd.Series(
    {
        "Argentina": 15,
        "France": 16,
        "Brazil": 8,
        "Croatia": 9,
        "Morocco": 6,
        "England": 13,
        "Portugal": 12,
        "Netherlands": "DNF",   # <- bad value, should have been a number
    },
    name="goals_scored",
)

print(goals)
```

```python
try:
    total = goals.sum()
    print(total)
except TypeError as e:
    goals = pd.to_numeric(goals,errors = "coerce")
    print(goals)

print(goals.sum())
```
