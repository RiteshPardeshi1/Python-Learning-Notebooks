# Q1. Count Uppercase and Lowercase Letters
Problem

Write a Python program that takes a sentence from the user and counts:

Number of uppercase letters
Number of lowercase letters

Ignore numbers and special characters.

```python
text = input("enter a sentence")
# take input
# two counters
upper = 0
lower = 0

for i in text:
    if "A" <= i <= "Z":
        upper += 1
    elif "a" <= i <= "z":
        lower += 1
print("uppercase", upper)
print("Lowercase", lower)
```

```python
text = input("enter a sentence")
# take input
# two counters
upper = 0
lower = 0

for i in text:
    if i.isupper():
        upper += 1
    elif i.islower():
        lower += 1
print("uppercase", upper)
print("Lowercase", lower)
```

# ✅ Q2. Count Digits and Alphabets
Problem

Count how many letters and digits are present.

```python
text = input("Enter a sentence")

digit = 0
alphabets = 0

for i in text:
    if i.isdigit():
        digit += 1
    elif i.isalpha():
        alphabets += 1
print("digit", digit)
print("alphabet", alphabets)
```

# ✅ Q3. Count Even and Odd Numbers
Problem
numbers = [10,15,8,7,20]

```python
numbers = [10,15,8,7,20]
even = 0 
odd = 0
```

```python
for num in numbers:
    if num % 2 == 0:
        even += 1
    else:
        odd += 1
print(even)
print(odd)
```
