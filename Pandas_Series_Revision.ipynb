## Q11 Lambda Function 

expenses = pd.Series({"rent": 12000, "food": 5000, "travel": 2000, "shopping": 3000})


````python
expenses = pd.Series({"rent": 12000, "food": 5000, "travel": 2000, "shopping": 3000})
````

````python

## 11.1. Use a lambda function with `.apply()` to create a new Series that labels each expense `"High"` if it is greater than 4000, otherwise `"Low"`.


classify = expenses.apply(lambda x: "High" if x > 4000 else "Low")
classify
````

````python
## 2. Using `functools.reduce`, find the total of all values in the `expenses` Series from Q11 — without calling `.sum()`.

sum = reduce(lambda x,y: x+ y,expenses)
````
