⭐ Q13. Count Frequency of Characters
Problem

Given a string, count how many times each character appears.

```python
text = 'banana'
frequency = {}

for x in text:
    if x in frequency:
        frequency[x] += 1
    else:
        frequency[x] = 1
print(frequency)
```
